```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)

OUT_FEAT_PATH = "../data/processed/financial_database_features.parquet"

df = pd.read_parquet(OUT_FEAT_PATH)

# Filter for tesla data
aapl_df = df[df['ticker'] == 'TSLA'].copy()

# Assume there's a recession indicator column, or create one based on common recession periods
# If your dataframe has a 'recession' column:
recession_periods = aapl_df[aapl_df['macro_regime'] == "recession"]

# Create figure with subplots
fig, axes = plt.subplots(2, 1, figsize=(14, 8))

# Plot 1: Price during recessions
axes[0].scatter(recession_periods['date'], recession_periods['close'], alpha=0.6, s=20)
axes[0].set_title('AAPL Stock Price During Recessions')
axes[0].set_ylabel('Close Price')
axes[0].set_xlabel('Date')
axes[0].grid(True, alpha=0.3)

# Plot 2: Price during expansions
expansion_periods = aapl_df[aapl_df['macro_regime'] == "expansion"]
axes[1].scatter(expansion_periods['date'], expansion_periods['close'], alpha=0.6, s=20, color='orange')
axes[1].set_title('AAPL Stock Price During Expansions')
axes[1].set_ylabel('Close Price')
axes[1].set_xlabel('Date')
axes[1].grid(True, alpha=0.3)
```
